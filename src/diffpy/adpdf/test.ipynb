```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = np.load("BaBiO3_adpdf_legacy.npz")

# List all arrays stored in the file
print("Keys in BaBiO3_adpdf_legacy.npz:", data.files)

print(data["r"])  # Print the radial distances
pdf = data["g_total"]
adpdf = data["g_ad"]
nonadpdf = data["g_nonad"]
r = data["r"][:413]
print(r[-1])

plt.figure(figsize=(10, 5))
plt.plot(r, pdf[:413], label="Total PDF")
plt.plot(r, adpdf[:413], label="AD PDF")
# plt.plot(r, nonadpdf[:413], label="Non-AD PDF")
plt.legend()
plt.show()
```

```python
%matplotlib ipympl
import matplotlib.pyplot as plt
import numpy as np

# Load the ADPDF.npz file
data = np.load("../ADPDFo.npz")

# List all arrays stored in the file
print("Keys in ADPDFo.npz:", data.files)

opdf = data["total-pdf"]
oBiBi = data["Bi-Bi-pdf"]
oBiBa = data["Bi-Ba-pdf"]
oBiO = data["Bi-O-pdf"]
oBi = 0.509 * oBiBi + 0.147 * oBiO + 0.344 * oBiBa
onBi = (opdf - 0.509 * oBi) / 0.491
# plt.plot(data["r"], opdf, label="Total PDF")
# plt.plot(data["r"], BaBi, label="Ba-Bi PDF")
# plt.plot(data["r"], BaBa, label="Ba-Ba PDF")
# plt.plot(data["r"], BaO, label="Ba-O PDF")
# plt.plot(data["r"], oBi, label="Bi PDF")
# plt.plot(r, pdf[:413], label="Total PDF")
# plt.plot(r, adpdf[:413], label="Non-Bi PDF")
plt.plot(data["r"], onBi, label="Bi PDF")
plt.plot(r, nonadpdf[:413], label="Non-Bi PDF")

plt.legend()
plt.show()
```
